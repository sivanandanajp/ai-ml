```python
from collections import deque
import heapq
```

```python
graph = {
    'S': [('A', 1), ('B', 4)],
    'A': [('B', 2), ('D', 5), ('C', 1)],
    'B': [('D', 2)],
    'C': [('D', 1), ('E', 3)],
    'D': [('G', 1)],
    'E': [('G', 2)],
    'G': []
}

def heuristic(node, goal):
    values = {
        'S': 6,
        'A': 4,
        'B': 5,
        'C': 3,
        'D': 2,
        'E': 1,
        'G': 0
    }
    return values.get(node, 0)

def greedy_best_first(graph, start, goal, heuristic_fn):
    if start == goal:
        return [start]

    pq = [(heuristic_fn(start, goal), start)]
    visited = set()
    parent = {start: None}

    while pq:
        _, current = heapq.heappop(pq)

        if current in visited:
            continue
        visited.add(current)

        if current == goal:
            path = []
            node = current
            while node is not None:
                path.append(node)
                node = parent[node]
            return path[::-1]

        for neighbor, _ in graph.get(current, []):
            if neighbor in visited:
                continue
            if neighbor not in parent:
                parent[neighbor] = current
            heapq.heappush(pq, (heuristic_fn(neighbor, goal), neighbor))

    return []

print('Greedy Best-First:', greedy_best_first(graph, 'S', 'G', heuristic))
```

```python
def a_star(graph, start, goal, heuristic_fn):
    if start == goal:
        return [start], 0

    pq = [(0, start)]
    g_score = {start: 0}
    parent = {start: None}

    while pq:
        _, current = heapq.heappop(pq)

        if current == goal:
            path = []
            node = current
            while node is not None:
                path.append(node)
                node = parent[node]
            return path[::-1], g_score[current]

        for neighbor, edge_cost in graph.get(current, []):
            tentative_g = g_score[current] + edge_cost
            if tentative_g < g_score.get(neighbor, float('inf')):
                parent[neighbor] = current
                g_score[neighbor] = tentative_g
                f_score = tentative_g + heuristic_fn(neighbor, goal)
                heapq.heappush(pq, (f_score, neighbor))

    return [], float('inf')

path, cost = a_star(graph, 'S', 'G', heuristic)
print('A* path:', path)
print('A* cost:', cost)
```
