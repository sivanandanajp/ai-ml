```python
import heapq

graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2), ('E', 5)],
    'C': [('F', 3)],
    'D': [('G', 1)],
    'E': [('G', 2)],
    'F': [('G', 2)],
    'G': []
}

def ucs(start, goal):
    queue = [(0, start, [start])]
    visited = set()

    while queue:
        cost, current, path = heapq.heappop(queue)

        if current == goal:
            return cost, path

        if current in visited:
            continue

        visited.add(current)

        for next, weight in graph[current]:
            if next not in visited:
                heapq.heappush(queue, (cost + weight, next, path + [next]))

cost, path = ucs('A', 'G')

print("Cost:", cost)
print("Path:", path)
```
