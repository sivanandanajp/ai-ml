```python

# STEP 1: Import libraries
import pandas as pd
from google.colab import files
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# STEP 2: Upload dataset
uploaded = files.upload()
filename = next(iter(uploaded))
df = pd.read_csv(filename)
```

```python
print("Dataset:")
print(df.head())
print("\nColumns:", df.columns.tolist())

# STEP 3: Remove unnecessary columns
df = df.drop(columns=["id", "Unnamed: 32"], errors="ignore")

# STEP 4: Remove diagnosis column from clustering features
X = df.drop(columns=["diagnosis"], errors="ignore")

# Keep numeric columns only
X = X.select_dtypes(include="number")

# Handle missing values
X = X.fillna(X.mean())

# Remove constant columns
X = X.loc[:, X.nunique() > 1]

# STEP 5: Scale data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# STEP 6: Apply K-Means
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
clusters = kmeans.fit_predict(X_scaled)

# STEP 7: Display clusters
df["Cluster"] = clusters

print("\nCluster assignments:")
print(df[["Cluster"]].head(10))

print("\nCluster counts:")
print(df["Cluster"].value_counts().sort_index())

# STEP 8: Evaluate clustering
print("\nSilhouette Score:",
      round(silhouette_score(X_scaled, clusters), 3))
```
